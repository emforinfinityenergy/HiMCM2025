## CRITIC

```python
def critic_weight(X: np.ndarray):
    X = np.array(X, dtype=float)

    X_min = X.min(axis=0)
    X_max = X.max(axis=0)
    norm_X = (X - X_min) / (X_max - X_min + 1e-12)

    std = np.std(norm_X, axis=0, ddof=1)
    corr = np.corrcoef(norm_X, rowvar=False)
    conflict = (1 - corr).sum(axis=1)
    contrast = std * conflict
    weights = contrast / contrast.sum()

    return weights, contrast
```
