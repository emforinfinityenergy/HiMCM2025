C# Combined Weighting

## Setup

```python
import numpy as np

entropy_weights = np.array([0.83525207, 0.02826386, 0.13648407])
entropy = np.array([0.89571899, 0.99647126, 0.98296])

critic_weight = np.array([0.32504345, 0.35262539, 0.32233116])
contrast = np.array([0.82154075, 0.89125357, 0.8146855])
```

## Calculation

```python
s1 = np.sum(1 - entropy)
s2 = np.sum(contrast)
delta = s1 / (s1 + s2)
print(delta * entropy_weights[0] + (1 - delta) * critic_weight[0])
concat_weight = np.array([delta * entropy_weights[i] + (1 - delta) * critic_weight[i] for i in range(3)])
```
