## Entropy Weighting

```python
def entropy_weight(X: np.ndarray):
    X = np.array(X, dtype=float)
    eps = 1e-12

    P = X / (X.sum(axis=0) + eps)

    n = X.shape[0]
    entropy = - (P * np.log(P + eps)).sum(axis=0) / np.log(n)

    d = 1 - entropy
    weights = d / d.sum()

    return weights, entropy
```
