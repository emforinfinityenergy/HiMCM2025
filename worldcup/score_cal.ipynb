# Calculating Indicator Scores

## Setup

```python
import numpy as np

city = np.array(["Doha", "London", "Barcelona", "Los Angeles"])

reg = lambda x: (1 - np.tanh(40 * x - 2)) / 2
```

```python
baseline = np.array([1268025850, 43, 2129.54, 3630000])

temp = np.array([30, 16.1, 24, 22.5])
loc = np.array([2.2, 1.4, 0.8, 3.9])
trans = np.array([43, 45, 20, 70])
energy = np.array([99.7, 53.9, 56, 62])
landfill = np.array([95, 5, 26, 40])
airport = np.array([9.7, 9.4, 11.1, 2.5])
res = np.array([8.45350568e10, 686847336, 5.81178515e10, 1.40011188e11])
neutralization = np.array([191625 / 365, 1724625 / 365, 1053600 / 365, 766500 / 365])

TIME_SPAN = 30
```

## Calculation

```python
# water
alpha = 3.5
water = reg(alpha * temp / temp[0] * baseline[0] / (res * TIME_SPAN))
print(list(water))

# waste
beta = 0.2
waste = (beta * temp[2] * landfill / 100 + (1 - beta) * temp[1] * landfill / 100) * -1
print(list(waste))

# gas
gamma = 0.79
sigma = 1.1
epsilon = 1.4
eta = 1.03
emission = reg(gamma * loc / loc[0] * sigma * trans / trans[0] * epsilon * energy / energy[0] * eta * airport / airport[0] / (neutralization * TIME_SPAN))
print(gamma * loc / loc[0] * sigma * trans / trans[0] * epsilon * energy / energy[0] * eta * airport / airport[0] / neutralization)
print(list(emission))
```

## Concat

```python
res = []
for i in range(len(water)):
    res.append([water[i], waste[i], emission[i]])

print(res)
```
