# VIKOR

```python
import numpy as np

entropy_weights = np.array([4.57411150e-01, 5.42588848e-01, 1.83540696e-09])
critic_weight = np.array([0.5158417 , 0.24884219, 0.23531612])
combined_weight = np.array([0.50819318, 0.28729345, 0.20451338])

decision = [[np.float64(0.9793667027775559), np.float64(-16.796), np.float64(0.981901022343932)], [np.float64(0.005234792766338758), np.float64(-0.8840000000000001), np.float64(0.9820094313076377)], [np.float64(0.978898646189329), np.float64(-4.596800000000001), np.float64(0.9820115670995451)], [np.float64(0.9808591443139765), np.float64(-7.072000000000001), np.float64(0.9820007858836555)]]
```

```python
def vikor(X: np.ndarray, weights: np.ndarray, benefit: np.ndarray, v: float = 0.5):
    X = np.array(X, dtype=float)
    weights = weights / weights.sum()
    m, n = X.shape

    f_best = np.where(benefit, X.max(axis=0), X.min(axis=0))
    f_worst = np.where(benefit, X.min(axis=0), X.max(axis=0))

    S = np.zeros(m)
    R = np.zeros(m)

    for i in range(m):
        diff = weights * (f_best - X[i]) / (f_best - f_worst + 1e-12)
        S[i] = diff.sum()
        R[i] = diff.max()

    S_min, S_max = S.min(), S.max()
    R_min, R_max = R.min(), R.max()

    Q = v * (S - S_min) / (S_max - S_min + 1e-12) + \
        (1 - v) * (R - R_min) / (R_max - R_min + 1e-12)

    ranking = np.argsort(Q)

    # return S, R, Q, ranking
    return Q, ranking
```

```python
print("Entropy", vikor(decision, entropy_weights, np.array([True, True, True]), 0.7))
print("CRITIC", vikor(decision, critic_weight, np.array([True, True, True]), 0.7))
print("Combined", vikor(decision, combined_weight, np.array([True, True, True]), 0.7))
```
