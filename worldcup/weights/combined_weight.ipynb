# Combined Weighting

## Setup

```python
import numpy as np

entropy_weights = np.array([4.57411150e-01, 5.42588848e-01, 1.83540696e-09])
entropy = np.array([0.8004745 , 0.76331948, 1.        ])

critic_weight = np.array([0.5158417 , 0.24884219, 0.23531612])
contrast = np.array([1.49396535, 0.72068933, 0.68151553])
```

## Calculation

```python
s1 = np.sum(1 - entropy)
s2 = np.sum(contrast)
delta = s1 / (s1 + s2)
print(delta * entropy_weights[0] + (1 - delta) * critic_weight[0])
concat_weight = np.array([delta * entropy_weights[i] + (1 - delta) * critic_weight[i] for i in range(3)])
```
