# reg(x) Plotter
This notebook plots $\text{reg}(x)$ and $\dfrac{\text{d}(\text{reg}(x))}{\text{d}x}$.
Note that $\text{reg}(x)=\dfrac{1+\tanh(4x-2)}{2}$

```python
import numpy as np
import matplotlib.pyplot as plt

reg = lambda x: (1 + np.tanh(4 * x - 2)) / 2
d_reg = lambda x: 2 * 1 / np.cosh(4 * x - 2) / np.cosh(4 * x - 2)

plt.figure(dpi=300, figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$\text{reg}(x)$")
plt.scatter(0, reg(0), marker="+", color="black", zorder=10)
plt.annotate(f'$(0, {round(reg(0), 5)})$',
             xy=(0, reg(0)), xycoords='data',
             xytext=(-20, +15), textcoords='offset points', fontsize=8,
             )
plt.scatter(1, reg(1), marker="+", color="black", zorder=10)
plt.annotate(f'$(1, {round(reg(1), 5)})$',
             xy=(1, reg(1)), xycoords='data',
             xytext=(-60, 0), textcoords='offset points', fontsize=8,
             )
plt.plot(np.linspace(-0.1, 1.1, 100000), reg(np.linspace(-0.1, 1.1, 100000)))

plt.subplot(1, 2, 2)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$\frac{\text{d}(\text{reg}(x))}{\text{d}x}$")
plt.scatter(0.5, d_reg(0.5), marker="+", color="black", zorder=10)
plt.annotate(f'$(0, {round(d_reg(0.5), 5)})$',
             xy=(0.5, d_reg(0.5)), xycoords='data',
             xytext=(-40, 0), textcoords='offset points', fontsize=8,
             )

plt.plot(np.linspace(-0.1, 1.1, 100000), d_reg(np.linspace(-0.1, 1.1, 100000)))

plt.show()
```
