# norm(x) Plotter
This notebook plots $\text{norm}(x)$ and $\dfrac{\text{d}(\text{norm}(x))}{\text{d}x}$.
Note that $\text{norm}(x)=\dfrac{1+\tanh(4x-2)}{2}$

```python
import numpy as np
import matplotlib.pyplot as plt

norm = lambda x: (1 + np.tanh(4 * x - 2)) / 2
d_norm = lambda x: 2 * 1 / np.cosh(4 * x - 2) / np.cosh(4 * x - 2)

plt.figure(dpi=300, figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$\text{norm}(x)$")
plt.scatter(0, norm(0), marker="+", color="black", zorder=10)
plt.annotate(f'$(0, {round(norm(0), 5)})$',
         xy=(0, norm(0)), xycoords='data',
         xytext=(-20, +15), textcoords='offset points', fontsize=8,
)
plt.scatter(1, norm(1), marker="+", color="black", zorder=10)
plt.annotate(f'$(1, {round(norm(1), 5)})$',
         xy=(1, norm(1)), xycoords='data',
         xytext=(-60, 0), textcoords='offset points', fontsize=8,
)
plt.plot(np.linspace(-0.1, 1.1, 100000), norm(np.linspace(-0.1, 1.1, 100000)))

plt.subplot(1, 2, 2)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$\frac{\text{d}(\text{norm}(x))}{\text{d}x}$")
plt.scatter(0.5, d_norm(0.5), marker="+", color="black", zorder=10)
plt.annotate(f'$(0, {round(d_norm(0.5), 5)})$',
         xy=(0.5, d_norm(0.5)), xycoords='data',
         xytext=(-40, 0), textcoords='offset points', fontsize=8,
)

plt.plot(np.linspace(-0.1, 1.1, 100000), d_norm(np.linspace(-0.1, 1.1, 100000)))

plt.show()
```
